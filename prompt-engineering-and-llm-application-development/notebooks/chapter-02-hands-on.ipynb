## Role-Prompting

Role prompting is a technique where you instruct the LLM to adopt a specific **persona, profession, or character** while performing a task.

Instead of just giving the model a task, you define *who the model should be* - this influences the tone, style, level of detail, and focus of the response.

### Key Benefit:
The same task can produce very different outputs depending on the assigned role, allowing you to control expertise level, perspective, and communication style.

```python
# Example 1: Explaining API Rate Limiting - Basic vs Role-Based

basic_prompt = "Explain API rate limiting."

role_prompt = """
You are a technical writer creating documentation for developers.
Explain API rate limiting clearly and concisely.
Outline why it is necessary and common ways it is implemented.
"""

print("=== BASIC PROMPT ===")
response = openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[{"role": "user", "content": basic_prompt}],
    temperature=0.7,
    max_tokens=200
)
display(Markdown(response.choices[0].message.content))

print("\n=== ROLE-BASED PROMPT (Technical Writer) ===")
response = openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[{"role": "user", "content": role_prompt}],
    temperature=0.7,
    max_tokens=200
)
display(Markdown(response.choices[0].message.content))
```

```python
# Example 2: Same Topic, Different Roles - Containerization

topic = """
Discuss the benefits of containerization in software deployment.
"""

devops_prompt = f"""
You are a seasoned DevOps Engineer with 10 years of experience.
{topic}
Focus on deployment speed, environment consistency, and infrastructure efficiency.
"""

marketing_prompt = f"""
You are a Marketing Manager at a software company.
{topic}
Focus on business benefits, cost savings, and competitive advantages.
"""

print("=== DEVOPS ENGINEER PERSPECTIVE ===")
response = openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[{"role": "user", "content": devops_prompt}],
    temperature=0.6,
    max_tokens=250
)
display(Markdown(response.choices[0].message.content))

print("\n=== MARKETING MANAGER PERSPECTIVE ===")
response = openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[{"role": "user", "content": marketing_prompt}],
    temperature=0.6,
    max_tokens=250
)
display(Markdown(response.choices[0].message.content))
```

```python
# Example 3: Explaining Complex Concepts - Audience Level Matters

topic = "Explain machine learning model overfitting"

beginner_role = f"""
You are a patient coding instructor teaching beginners.
{topic}
Use simple language, avoid jargon, and include a relatable analogy.
Keep your explanation under 100 words.
"""

expert_role = f"""
You are a machine learning researcher.
{topic}
Include technical details about regularization, bias-variance tradeoff, and cross-validation.
Target an audience familiar with statistical concepts.
"""

print("=== BEGINNER-FRIENDLY EXPLANATION ===")
response = openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[{"role": "user", "content": beginner_role}],
    temperature=0.5,
    max_tokens=200
)
display(Markdown(response.choices[0].message.content))

print("\n=== EXPERT-LEVEL EXPLANATION ===")
response = openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[{"role": "user", "content": expert_role}],
    temperature=0.5,
    max_tokens=300
)
display(Markdown(response.choices[0].message.content))
```

```python
# Example 4: Role-Based Code Review

code_sample = """
def calculate_average(numbers):
    total = 0
    for num in numbers:
        total = total + num
    average = total / len(numbers)
    return average
"""

strict_reviewer = f"""
You are a strict code reviewer focused on best practices and performance.
Review the following Python function and provide critical feedback.
Suggest improvements for readability, efficiency, and adherence to Python conventions.

Code:
{code_sample}
"""

supportive_reviewer = f"""
You are a senior mentor reviewing a junior developer's code.
Review the following Python function with constructive feedback.
Acknowledge what's good about it, then suggest improvements in an encouraging way.
Help them learn, don't criticize.

Code:
{code_sample}
"""

print("=== STRICT CODE REVIEWER ===")
response = openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[{"role": "user", "content": strict_reviewer}],
    temperature=0.5,
    max_tokens=300
)
display(Markdown(response.choices[0].message.content))

print("\n=== SUPPORTIVE MENTOR REVIEWER ===")
response = openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[{"role": "user", "content": supportive_reviewer}],
    temperature=0.5,
    max_tokens=300
)
display(Markdown(response.choices[0].message.content))
```

## Structuring Output Formats

```python
# Example 5: Combining Structured Output with Validation
print("=" * 60)
print("Example 5: Structured Output with Validation")
print("=" * 60)

def validate_contact_json(data):
    """Validate that the JSON has required fields"""
    required_fields = ["name", "email", "company"]
    if not isinstance(data, dict):
        return False, "Output is not a dictionary"
    
    for field in required_fields:
        if field not in data:
            return False, f"Missing required field: {field}"
    
    return True, "Valid"

text = "Contact Maria Garcia from CloudServices Ltd. at maria@cloudservices.io for partnership inquiries."

prompt = f"""Extract contact information from the text.
Return a JSON object with: name, email, company, phone (null if missing).

Text: "{text}"

Return ONLY the JSON object."""

response = openai.messages.create(
    model="claude-3-5-sonnet-20241022",
    max_tokens=1024,
    messages=[{"role": "user", "content": prompt}]
)

json_output = response.content[0].text
print(f"\nRaw Response:\n{json_output}\n")

try:
    contact = json.loads(json_output)
    is_valid, message = validate_contact_json(contact)
    
    print(f"Validation: {message}")
    print(f"Parsed Data:\n{json.dumps(contact, indent=2)}\n")
    
    if is_valid:
        print("✓ JSON is valid and can be used in downstream applications")
    else:
        print("✗ JSON validation failed")
except json.JSONDecodeError as e:
    print(f"✗ Error parsing JSON: {e}\n")
```

```python
# Example 4: Using Schema Definition with Few-Shot Examples
print("=" * 60)
print("Example 4: Schema Definition with Few-Shot Examples")
print("=" * 60)

prompt = """Extract product information and return as JSON.

Schema:
{
  "product_name": "string",
  "price": "number or null",
  "currency": "string",
  "in_stock": "boolean",
  "features": ["string array"]
}

Examples:
Input: "The new MacBook Pro 16 inch costs $3,499, available now, with M4 chip, 16GB RAM, and 512GB SSD"
Output: {"product_name": "MacBook Pro 16 inch", "price": 3499, "currency": "USD", "in_stock": true, "features": ["M4 chip", "16GB RAM", "512GB SSD"]}

Input: "The vintage leather briefcase - no pricing info, limited stock, features genuine leather and brass handles"
Output: {"product_name": "vintage leather briefcase", "price": null, "currency": null, "in_stock": false, "features": ["genuine leather", "brass handles"]}

Now extract from this:
Input: "The wireless noise-cancelling headphones, priced at €249.99, in stock, with 30-hour battery, Bluetooth 5.0, and active noise cancellation"

Return ONLY the JSON object."""

response = openai.messages.create(
    model="claude-3-5-sonnet-20241022",
    max_tokens=1024,
    messages=[{"role": "user", "content": prompt}]
)

json_output = response.content[0].text
print(f"\nRaw Response:\n{json_output}\n")

try:
    product_data = json.loads(json_output)
    print(f"Parsed Product Data:\n{json.dumps(product_data, indent=2)}\n")
except json.JSONDecodeError as e:
    print(f"Error parsing JSON: {e}\n")
```

```python
# Example 3: Markdown Formatted Output
print("=" * 60)
print("Example 3: Markdown Formatted Output")
print("=" * 60)

prompt = """Summarize the benefits of Large Language Models for customer support.

Format your response as Markdown with:
- An H2 heading
- A bulleted list of benefits
- Bold key concepts

Do not include any text outside the Markdown structure."""

response = openai.messages.create(
    model="claude-3-5-sonnet-20241022",
    max_tokens=1024,
    messages=[{"role": "user", "content": prompt}]
)

markdown_output = response.content[0].text
print(f"\nMarkdown Output:\n")
print(markdown_output)
print()
```

```python
# Example 2: Multiple Records in JSON Array
print("=" * 60)
print("Example 2: Multiple Records in JSON Array")
print("=" * 60)

emails_text = """
1. Contact: Alice Johnson, Email: alice@example.com, Company: StartupXYZ
2. Contact: Bob Chen, Email: bob.chen@enterprise.com, Company: BigCorp
3. Contact: Carol Williams, Phone: 206-555-9876, Company: LocalBiz
"""

prompt = f"""Extract all contacts from the text and return a JSON array.
Each object should have: name, email (or null), phone (or null), company.

Text:
{emails_text}

Return ONLY the JSON array, no other text."""

response = openai.messages.create(
    model="claude-3-5-sonnet-20241022",
    max_tokens=1024,
    messages=[{"role": "user", "content": prompt}]
)

json_output = response.content[0].text
print(f"\nRaw Response:\n{json_output}\n")

try:
    contacts = json.loads(json_output)
    print(f"Parsed Contacts:\n{json.dumps(contacts, indent=2)}\n")
    print(f"Number of contacts extracted: {len(contacts)}\n")
except json.JSONDecodeError as e:
    print(f"Error parsing JSON: {e}\n")
```
